## Netflix 90s Movie Analysis

Netflix started in 1997 as a DVD rental service and has since grown into one of the world's largest entertainment platforms. With thousands of titles spanning decades, it's a great dataset for exploring trends in cinema history.

In this notebook, I explore Netflix movies from the **1990s** — analyzing their durations, genres, and identifying short-form Action films from the decade.

## The data
### **netflix_data.csv**
| Column | Description |
|--------|-------------|
| `show_id` | The ID of the show |
| `type` | Type of show |
| `title` | Title of the show |
| `director` | Director of the show |
| `cast` | Cast of the show |
| `country` | Country of origin |
| `date_added` | Date added to Netflix |
| `release_year` | Year of Netflix release |
| `duration` | Duration of the show in minutes |
| `description` | Description of the show |
| `genre` | Show genre |

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv("netflix_data.csv")
netflix_df.head()
```

```python
# Quick overview of the dataset
print(f"Dataset shape: {netflix_df.shape}")
print(f"\nColumns: {list(netflix_df.columns)}")
print(f"\nRelease year range: {netflix_df['release_year'].min()} – {netflix_df['release_year'].max()}")
```

```python
# Filter to 1990s movies only
cond = np.logical_and(netflix_df['release_year'] >= 1990, netflix_df['release_year'] <= 1999)
mov_90s = netflix_df[cond]
mov_dur = mov_90s['duration']

# Plot distribution of movie durations
plt.figure(figsize=(10, 5))
plt.hist(mov_dur, bins=max(mov_dur), color='steelblue', edgecolor='white')
plt.xlabel('Duration (minutes)')
plt.ylabel('Number of Movies')
plt.title('Distribution of 90s Netflix Movie Durations')
plt.tight_layout()
plt.show()

print(f"Longest 90s movie on Netflix: {max(mov_dur)} minutes")
print(f"Total 90s movies: {len(mov_90s)}")
```

```python
# Short Action movies from the 90s (under 90 minutes)
short_movie_action = mov_90s[mov_90s['genre'] == 'Action']
short_movie_action_l90 = short_movie_action[short_movie_action['duration'] < 90]

print(f"Short 90s Action movies (under 90 min): {len(short_movie_action_l90)}")
short_movie_action_l90[['title', 'director', 'duration', 'country']].reset_index(drop=True)
```
